1. В первый день спортсмен пробежал x километров, а затем он каждый день увеличивал пробег на 10% от предыдущего значения. По данному числу y определите номер дня, на который пробег спортсмена составит не менее y километров. Программа получает на вход действительные числа x и y и должна вывести одно натуральное число.

````python
x = float(input())
y = float(input())
yy = x
counter = 1
while (yy - y) < 0.1 * yy:#более строгое, чем yy<y, на 5-й пробегает 1.4531x
    yy += 0.1*y
    counter += 1
print (counter)
````

2. Вывести все простые числа в заданном интервале Например, [1,27]

````python
a = int(input())
b = int(input()) #пусть ввод пока такой, должно быть map(split)
for i in range (a,b+1):#вторая граница не включается
    counter = 0
    for j in range (2, int(round(i/2)+1)):#i//2+1 - то же самое, но по умному вообще до корня+1
        if i % j == 0: 
            counter += 1 #или через булевую переменную
            break #выход из цикла по первому делителю 
    if counter == 0:
        print (i)
````

**Задача со звездочкой ***

Дан список слов, необходимо выделить общую начальную часть (с максимальной длиной) всех слов в списке, если такой нет, вывести пустую строку.

Например, ['радуга', 'радость', 'радио', 'радар'], результат 'рад', или ['радуга', 'радость', 'радио', 'маяк'], результат ''

````python
import ast
words = ast.literal_eval(input()) #сам разберёт написанное в список, без "split", но нужны ['']
gen = words[0] #первое слово предполагается общей начальной частью
for word in words[1:]: #срез списка, начиная со второго элемента
    while not word.startswith(gen): #строковый метод, который проверяет начало строки (true/false)
        gen = gen[:-1] #по одной букве сокращается строка
        if gen == "": break
    if gen == "": break #выход по пустой общей части
print (gen)
````
